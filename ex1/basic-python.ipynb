## Batch mode

Batch mode is how Python is typically executed in practice, and is more similar to how more traditional, compiled languages function. In batch mode, python syntax is written to a file which can be executed. In this folder, there is a file called "first_script.py". If your terminal is in this directory, then typing 

```bash
python first_script.py
```

will do some computations and write them to the console. Open up the file in your favorite text editor to look inside. 

This file shows some great syntax to be aware of in Python. If you are not used to Python code, you might reel at the fact that indentation is used to encapsulate scopes, rather than more traditional curly braces. This might be difficult to get used to, but it ensures you quickly understand the depth level a particular line is executed on.

First to note, single line comments are specified by 
```python
#This is a comment
```
while multi-line comments by 
```python
"""
This is a multi-line 
comment.
"""
```
These lines are not read by the interpreter, and are used for human-readable communication.

Second, all code occurs inside a strange if statement:
```python
if __name__ == "__main__":
```
In essence, this line of code checks that you intended to run the file as a script. While not strictly necessary, it is good practice to include it since all lines of code at the outer-most level of the script will be executed whenever the file is loaded (like for instance, if you want to import a function from another file).

Also note that the `print` function takes multiple arguments, and that the integer `result` is automatically cast to a string. 

As you might notice, functions in Python are declared with the following syntax: 
```python
def myfunction(x,y=2):
    return x + y
```
This simple function returns the result of adding its two parameters together. Python supports default values, notice how `y=2` is specificed in the argument. This means that if you call the function with only one parameter i.e. `myfunction(5)`, it will be treated the same as if you had called `myfunction(5,2)`. All default parameters must be specified *after* required parameters.

Lambda functions (anonymous functions) are also a valid way to write functions in Python, which can be a nice way to specify a one-line function or to pass a function as an argument. They are specified by 
```python
mylambda = lambda x: x + 2
```
and are called just like functions, `mylambda(5)` will return the value `7`.

One final note is the use of f-strings (formatted strings). Placing the decorator 'f' immediately before specifying a string tells the interpreter it is an f-string. In such an object, any value placed between curly brackets `{}` is evaluated, cast to a string, and inserted in place. This is very handy for debugging, or just writing clean code.

## Types 
Python types come in roughly three variants: 
- strings
- numbers
- collections

We have already looked a lot at strings, but it is worth mentioning they are both immutable (cannot be modified in-place) and iterable (can be looped over). 

Numbers are either `int` (integers), `float` (floating point), or `complex`. The interpreter will automatically cast int to float if needed (and floats to complex). 

Collections are the primary way you will interact with Python. For now, we cover only two: the list and dictionary. 

### List 
The list is similar to an array in other languages. It is an ordered collection of objects. They are specified with square brackets:
```python
onelist = [1,2,3,4,5]
```
or an empty list can be instantiated with the keyword
```python
newlist = list()
```


There are two key differences between a traditional array and a Python list. First, it does not have a fixed length as it is essentially a dynamic array styled data structure. 

````python
onelist = [1,2,3,4,5]
onelist.append(6)
onelist.append(7)
onelist.insert(3,10)

onelist
````

Secondly, it may contain values of different types:

````python
twolist = [1, "two", 3.5, 4+2j, onelist]

twolist
````

Note that lists can themselves contain lists (or other collections). However, mixing the types contained in a list is generally best avoided. 

### Dictionary
A dictionary is a key-value pair data structure (in fact, a hash map). Other languages call this a "map", since it maps some set of values to another in a one-to-one fashion. Dictionaries are specified with curly brackets: 
```python
onedictionary = {"class": "data science", 16: 32}

#or an empty dict 
emptydict = dict()
```
The "inputs" to the dictionary are called keys, while the "outputs" are called values. A key must be a hashable (immutable) object, while any object can be a value. 

## Flow control
The last of the basic syntax we will cover is flow control. There are generally two ways to do this, for loops and while loops.

### For loops
For loops in Python are more like a for-in loop in other languages. That is, they go through items in a collection one at a time, like rifling through papers in a filing cabinet. 

````python
example = [1,2,3,4,5]
for number in example:
    print(number)
````

For loops are specified with the syntax: `for [iterator] in [collection]:`
The value of `iterator` changes each execution of the loop, according to the next item in collection. Collections which can be used in a for loop are called "iterable".

If you are coming from other languages, you are probably used to using a loop index. This can be done using the built-in function, `range(x)`, which produces an iterable with items 0,1,..., x-1.

````python
example2 = ["Data", "science", "is cool."]
for index in range(3):
    print(index, example2[index])
````

However, Python designers don't recommend this. Often, a more elegant solution is to use the built-in `enumerate()` function, which takes an iterable as input and returns a pair of (index, value).

````python
example2 = ["Data", "science", "is cool."]
for index,value in enumerate(example2):
    print(index, value)
````

In a rather strange design decision, the preferred method to create lists is through the (in)famous `list comprehension`. It is mainly preferred due to it being efficient; large lists in Python are slow, especially if they are created through `.append` or `.insert` operations. 

List comprehensions use an in-line for loop in the following syntax

````python
lcomp = [x for x in range(10)]
lcomp
````

What this does is iterate through the collection `range(10)`, placing the value `x` at each iteration into the list. Of course, the first `x` need not be only a variable, but can be a more complicated expression which is then evaluated.

````python
lcomp2 = [2*x - 45 + i**2 for i,x in enumerate([2,4,6,8,10])]
lcomp2
````

### While loops

While loops work exactly as you would expect if you have programmed in other languages. They evaluate a condition on each iteration, and continue looping as long as this condition is true. 

````python
x = 0 
while x <= 10:
    print(x * 2)
    x += 2
````

The program above prints the double of each even number between 0 and 10. 

While loops should be used sparingly, as they are more difficult to read and control than for loops. However, they have the advantage of having less overhead and can be used when you don't know exactly how many times you need to loop.

## Problem 1: Count Odds

Write a function that returns how many odd numbers there are in a list. You can assume the list will contain only numbers.

````python
# your code
def count_odds(x):
   count = 0
   for valueArr in x:
       if valueArr % 2 == 1:
            count += 1
   return count
````

## Problem 2: Reversing a List Two Ways

Write two functions `reverse_element` and `reverse_lcomp` that both take a list as a parameter and reverse the order of its elements, returning a new list.

`reverse_element` should use element-wise processing in a loop

`reverse_lcomp` should use a list comprehension

````python
# your code
def reverse_element(x):
    reversedList = list()
    for number in x:
        reversedList.insert(0, number)
    return reversedList

def reverse_lcomp(x):
    return [number for number in x[::-1]]
````

````python
print(reverse_lcomp([3, 1, 4, 1, 5, 9]))
print(reverse_lcomp(["Base", "Foil", "Holographic", "Polychrome", "Negative"]))
print(reverse_lcomp([170, 171, 174, "Korra", "Asami", "Bolin", "Mako"]))
````

## Problem 3: Combining Two Lists

Write a function that combines two lists by alternatingly taking elements from the **back** of the first list and the **front** of the second list, e.g. [a,b,c], [1,2,3] → [c,1,b,2,a,3].
In case of two lists of different length, the result should be [a,b,c,d],[1,2]=>[d,1,c,2,b,a], i.e., the elements of the longer list are appended at the end.

````python
# your code
def combine_list(x,y):
    newList = list()
    x.reverse()
    index = 0
    for index in range(len(x)):
        newList.append(x[index])
        if index < len(y):
            newList.append(y[index])
    index += 1
    while index < len(y):
        newList.append(y[index])
        index += 1
    return newList
````

````python
# call your function with a couple of different lists and print the results here
print(combine_list(['w', 'x', 'y', 'z'], [1, 1, 2, 3, 5, 8]))
print()
print(combine_list([2, 7, 1, 8, 2, 8], ['delta', 'echo', 'foxtrot', 'golf']))
````

## Problem 4: Numbers to Digits in Two Ways

Write two functions that take a number and return a list of its digits in order, e.g., 789456 → [7, 8, 9, 4, 5, 6].

You need to implement two versions of the function: `num_to_digits_elem`, a version with element-wise processing, and `num_to_digits_lc`, a version using the list comprehension way.

````python
def num_to_digits_elem(number):
    digits = []
    for digit in str(number):
        digits.append(int(digit))
    return digits


def num_to_digits_lc(number):
    return [int(digit) for digit in str(number)]
````

## Problem 5: Pell Numbers

Write a function that takes a number `n` and an empty array that writes the first `n` [Pell numbers](https://en.wikipedia.org/wiki/Pell_number) into the empty array. Print the array. 

The Pell numbers can be generated via a relation where $Pell(x) = 2 * Pell(x-1) + Pell(x-2)$ with $Pell(1) = 1$ and $Pell(0) = 0$. Note that this sequence starts with $Pell(1)$ and $Pell(0)$ is just used in the relation. 

The sequence begins `1`, `(2*1 + 0) = 2`, `(2*2 + 1) = 5`, `(2*5 + 2) = 12`...

Thus, if you call `pell(4)`, it should thus produce `[1, 2, 5, 12]`.

````python
# your code
def pell(num, x):
    x.append(0)
    x.append(1)
    for index in range(1, num):
        x.append(2*x[index]+x[index-1])
    x.pop(0)
````

````python
# call your function with a couple of different n and print the results here
sequence = []
pell(17, sequence)
print(sequence)  
````
